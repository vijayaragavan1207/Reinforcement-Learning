```python
import numpy as np
import gymnasium as gym
```

```python
env = gym.make(
    "FrozenLake-v1",
    is_slippery=True
)

num_states = env.observation_space.n
num_actions = env.action_space.n

print("Number of states:", num_states)
print("Number of actions:", num_actions)
```

```python
gamma = 0.99
theta = 1e-6

# Initial value of all states
V = np.zeros(num_states)

print(V)
```

```python
iterations = 0

while True:
    delta = 0

    for state in range(num_states):
        action_values = []

        for action in range(num_actions):
            expected_value = 0

            for probability, next_state, reward, done in env.unwrapped.P[state][action]:
                future_value = 0 if done else V[next_state]

                expected_value += probability * (
                    reward + gamma * future_value
                )

            action_values.append(expected_value)

        old_value = V[state]
        V[state] = max(action_values)

        delta = max(
            delta,
            abs(old_value - V[state])
        )

    iterations += 1

    if delta < theta:
        break

print("Value Iteration completed.")
print("Iterations:", iterations)
```

```python
policy = np.zeros(num_states, dtype=int)

for state in range(num_states):
    action_values = []

    for action in range(num_actions):
        expected_value = 0

        for probability, next_state, reward, done in env.unwrapped.P[state][action]:
            future_value = 0 if done else V[next_state]

            expected_value += probability * (
                reward + gamma * future_value
            )

        action_values.append(expected_value)

    policy[state] = np.argmax(action_values)
```

```python
print("Calculated Grid Scores:")

print(
    np.round(
        V.reshape(4, 4),
        4
    )
)
```

```python
arrows = {
    0: "←",
    1: "↓",
    2: "→",
    3: "↑"
}

policy_arrows = np.array(
    [arrows[action] for action in policy]
).reshape(4, 4)

# Mark special FrozenLake positions
policy_arrows[0, 0] = "S"
policy_arrows[1, 1] = "H"
policy_arrows[1, 3] = "H"
policy_arrows[2, 3] = "H"
policy_arrows[3, 0] = "H"
policy_arrows[3, 3] = "G"

print("Optimal Directions (Policy Map):")
print(policy_arrows)
```
